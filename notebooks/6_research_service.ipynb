```python
import nest_asyncio
from config import settings
from research_service import ResearchService

# Apply patch to support nested event loops in Jupyter
nest_asyncio.apply()

# Initialize the service
service = ResearchService()

print(f"🔗 Render Deployment URL: {settings.RENDER_URL}")
print(f"🤖 Target Assistant ID : {settings.ASSISTANT_ID}")
print("✅ ResearchService initialized successfully!")
```

```python
print("--- TEST 1: Thread Session Creation ---")

try:
    thread_id = await service.start_session(user_id="notebook_test_user")
    print(f"✅ Created Thread ID: {thread_id}")
except Exception as e:
    print(f"❌ Session Creation Failed: {e}")
```

```python
print(f"--- TEST 2: Executing Research Pipeline [{thread_id}] ---\n")

test_query = "Summarize recent developments in solid-state batteries in 2 lines."

# Callbacks to handle SSE event streams
def handle_node_stage(node_name: str):
    print(f"\n📍 Stage Change -> Node: [{node_name}]")

def handle_token(token: str):
    print(token, end="", flush=True)

def handle_interrupt(payload: dict):
    print(f"\n\n❓ Interrupt Detected!")
    print(f"   Message  : {payload.get('message')}")
    print(f"   Questions: {payload.get('questions')}")

def handle_complete(final_state: dict):
    print("\n\n✅ Complete Event Triggered!")
    print(f"• Report Length: {len(final_state.get('final_report', ''))} characters")

try:
    print(f"🚀 Sending Query: '{test_query}'")
    
    await service.execute_research(
        thread_id=thread_id,
        user_query=test_query,
        on_node_stage=handle_node_stage,
        on_token=handle_token,
        on_interrupt=handle_interrupt,
        on_complete=handle_complete
    )
except Exception as e:
    print(f"\n❌ Pipeline Execution Failed: {e}")
```
